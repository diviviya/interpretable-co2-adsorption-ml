# 03 — Data Split and Preprocessing

This notebook prepares the audited CO₂ adsorption dataset for machine learning.

Objectives:
- Remove confirmed exact duplicate observations
- Assign stable row identifiers
- Define features and target
- Create reproducible train/validation/test splits
- Build leakage-safe preprocessing pipelines
- Handle missing feature values using training-set statistics
- Apply scaling only where required
- Preserve the untouched test set for final evaluation

```python
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_PATH = PROJECT_ROOT / "data" / "Processed_527_datapoints.csv"

df = pd.read_csv(RAW_DATA_PATH)

print("Raw dataset shape:", df.shape)
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows before removal:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after duplicate removal:", df.shape)
```

```python
df.insert(0, "row_id", np.arange(len(df)))

print(df[["row_id"]].head())
print("Number of observations:", len(df))
```

```python
TARGET_COLUMN = "CO2 uptake (mmol/g)"

FEATURE_COLUMNS = [
    "Surface Area (m2/g)",
    "Total Pore Volume(cm3/g)",
    "Micropore Volume (cm3/g)",
    "C (%)",
    "H (%)",
    "N (%)",
    "O (%)",
    "S (%)",
    "Temp (°C)",
    "Pressure (bar)"
]

X = df[FEATURE_COLUMNS].copy()
y = df[TARGET_COLUMN].copy()

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Number of features:", len(FEATURE_COLUMNS))
```

```python
RANDOM_STATE = 42
TEST_SIZE = 0.20

X_dev, X_test, y_dev, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

print("Development features:", X_dev.shape)
print("Test features:", X_test.shape)
print("Development target:", y_dev.shape)
print("Test target:", y_test.shape)
```

```python
VALIDATION_SIZE = 0.20

X_train, X_val, y_train, y_val = train_test_split(
    X_dev,
    y_dev,
    test_size=VALIDATION_SIZE,
    random_state=RANDOM_STATE
)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
train_indices = set(X_train.index)
val_indices = set(X_val.index)
test_indices = set(X_test.index)

print("Train ∩ Validation:", len(train_indices & val_indices))
print("Train ∩ Test:", len(train_indices & test_indices))
print("Validation ∩ Test:", len(val_indices & test_indices))
```

```python
print(
    "Total unique observations:",
    len(train_indices | val_indices | test_indices)
)
```

```python
split_assignments = pd.DataFrame({
    "row_id": df.loc[X.index, "row_id"],
    "split": "unassigned"
}, index=X.index)

split_assignments.loc[X_train.index, "split"] = "train"
split_assignments.loc[X_val.index, "split"] = "validation"
split_assignments.loc[X_test.index, "split"] = "test"

split_assignments = split_assignments.sort_values("row_id").reset_index(drop=True)

split_assignments.head()
```

```python
print(split_assignments["split"].value_counts())
```

```python
SPLIT_DIR = PROJECT_ROOT / "artifacts" / "splits"
SPLIT_DIR.mkdir(parents=True, exist_ok=True)

split_path = SPLIT_DIR / "split_assignments.csv"

split_assignments.to_csv(split_path, index=False)

print("Split assignments saved to:", split_path)
```

```python
print("Missing values in TRAIN:")
print(X_train.isna().sum())

print("\nMissing values in VALIDATION:")
print(X_val.isna().sum())

print("\nMissing values in TEST:")
print(X_test.isna().sum())
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

tree_preprocessor = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])
```

```python
from sklearn.preprocessing import StandardScaler

ridge_preprocessor = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
tree_preprocessor.fit(X_train)
```

```python
imputer = tree_preprocessor.named_steps["imputer"]

print("Training medians:")
for feature, median in zip(FEATURE_COLUMNS, imputer.statistics_):
    print(f"{feature}: {median}")
```

```python
X_train_tree = tree_preprocessor.transform(X_train)
X_val_tree = tree_preprocessor.transform(X_val)
X_test_tree = tree_preprocessor.transform(X_test)
```

```python
print("Train shape:", X_train_tree.shape)
print("Validation shape:", X_val_tree.shape)
print("Test shape:", X_test_tree.shape)
```

```python
print("Missing values after imputation:")
print(np.isnan(X_train_tree).sum())
print(np.isnan(X_val_tree).sum())
print(np.isnan(X_test_tree).sum())
```

```python
ridge_preprocessor.fit(X_train)
```

```python
X_train_ridge = ridge_preprocessor.transform(X_train)
X_val_ridge = ridge_preprocessor.transform(X_val)
X_test_ridge = ridge_preprocessor.transform(X_test)
```

```python
print("Train shape:", X_train_ridge.shape)
print("Validation shape:", X_val_ridge.shape)
print("Test shape:", X_test_ridge.shape)
```

```python
import json

split_config = {
    "random_state": RANDOM_STATE,
    "test_size": TEST_SIZE,
    "validation_size_within_development": VALIDATION_SIZE,
    "train_rows": len(X_train),
    "validation_rows": len(X_val),
    "test_rows": len(X_test)
}

config_path = SPLIT_DIR / "split_config.json"

with open(config_path, "w") as f:
    json.dump(split_config, f, indent=4)

print("Split configuration saved to:", config_path)
```

```python
feature_config = {
    "target": TARGET_COLUMN,
    "features": FEATURE_COLUMNS
}

feature_config_path = PROJECT_ROOT / "artifacts" / "feature_config.json"
feature_config_path.parent.mkdir(parents=True, exist_ok=True)

with open(feature_config_path, "w") as f:
    json.dump(feature_config, f, indent=4)

print("Feature configuration saved to:", feature_config_path)
```
